## Dictionary comprehension

```python
d = {}
```

```python
[x**2 for x in range(1,6)]
```

```python
{i: i**2 for i in range (1,6)}
```

```python
d = {}
for i in range(1,6):
    d[i] = i**2

d
```

```python
fruits = ['apple', 'banana', 'orange', 'grapes']
```

```python
d = {}
for fruit in fruits:
    d[fruit] = len(fruit)
d    
```

```python
{fruit: len(fruit) for fruit in fruits}
```

```python
a = ['a','b','c']
r = [1,2,3]
```

```python
{a[i]: r[i] for i in range(len(a)) }
```

```python
d = {}
for i in range(len(a)):
    d[a[i]] = r[i]

print (d)
 
```

```python
emp_info = {'names': ['Jitender', 'Priyanka', 'Navneet', 'Bikram', 'Adarsh'] ,
            'salaries' : [2300, 5600,1300,3400,123]
}
```

```python
# Finding hoghest salary in company 
max(emp_info['salaries'])
```

```python
sum = 0
for salary in emp_info['salaries']:
    sum += salary
average = sum/len(emp_info['salaries'])
print (f"Average salary of the employees is", average)
```

```python
# print the employee names earning above 1500

for salary in emp_info['salaries']:
    if salary > 1500:
        print (emp_info['names'][emp_info['salaries'].index(salary)])
         
```

```python
for i in range (len (emp_info['salaries'])):
    if emp_info['salaries'][i] > 1500:
        print (emp_info['names'][i])
```
