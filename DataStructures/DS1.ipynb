####  Addition 

```python
l = [1,2,3,45,6]

# append - Adds the element at the last of list
l.append(10)
print (l)

# extend - Adds the element of one list to another at last
l1 = [1,3,5,6,8]
l2 = [45,23,56]

l1.extend(l2)
print (l1)
l2.extend(l1)
print (l2)

# insert - adds a element at specific location or index
l1 = [2,35,67,87]
l1.insert(2,23)
print (l1)
l1.insert(4,"jitender")
print(l1)
l1.insert(3,[2,3,4])
print(l1)
```

```python
# extend - Adds the element of one list to another
l1 = [1,3,5,6,8]
l2 = [45,23,56]

# l1 = [1,3,5,6,8,45,23,56]
l2.extend(l1)
print(l2)
```

```python
l1 = [2,34,45]
l2 = [3,5,6]
l3 = [53,34,67]

l1.extend(l2)
l1.extend(l3)
l1
```

### Remove 

```python
fruits = ['apple', 'orange', 'kiwi', 'pineapple', 'grapes', 'orange', 'litchi']

print(fruits)
# remove - it removes the first occurence of the value
fruits.remove('orange')
print(fruits)

# pop - Removes by index and will return the value at that index

fruits.pop(2)
print(fruits)

#del  -- It also takes the index value from which the element is to be removed.
del fruits[2]
print (fruits)
```

```python
# Modify 
print (fruits)
fruits[2] = 'banana'
print (fruits)
```

```python
# Checking membership 
'apple' in fruits
```

```python
len(fruits)
```

### Inbuilt functions for lists

```python
l = [1,2,34,56]
l.reverse()
l
```

```python
# Sort
l = [3,1,6,4,8,2]
l.sort()
print(f"Sorted in ascending order which is by default: {l}")
l.sort (reverse = True)
print(f"Sorted in descending order: {l}")
```

```python
# copy()

l1 = l.copy()
print (l1)

# Copying using the list function
l2 = list(l1)
print (l2)
```

```python
# count - it gives us the count of a value occured in the list
l = [1,2,3,1,4,1,6,9,3]
l.count(3)
```
