```python
def voter(age):
    if age >= 18:
        print ("Eligible")
    else :
        print ("Not eligible")
```

```python
def add(*args):
    result = 0
    for num in args:
        result += num
    return result
```

```python
add(1,23,4)
```

```python
# nested loops
for i in range (5):
    for j in range (3):
        print (i,j)
```

```python
# nested functions
def outerfunction(x):
    def inner_function(y):
        return x+y
    return inner_function   
```

```python
sol = outerfunction(5)
```

```python
sol(4)
```
